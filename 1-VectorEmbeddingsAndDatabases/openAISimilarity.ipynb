## Cosine Similarity With OpenAI Embeddings

```python
## Example1- Finding the Similar Sentences
sentences = [
    "The cat sat on the mat.",
    "A feling rested on the Rug",
    "The dog played in the yard",
    "I love programming in Python.",
    "Python is my favourite Programing Language."
]
```

```python
import numpy as np
def cosine_similarity(vec1,vec2):
    """
    Cosine Similarity measures the angle between the two vectors
    - Result close to 1: Very Similar
    - Result close to 0: not Related
    - Result close to -1: Oppposite Message
    """
    dot_product = np.dot(vec1,vec2)
    norm_a = np.linalg.norm(vec1)
    norm_b = np.linalg.norm(vec2)
    return dot_product/((norm_a)*(norm_b))
```

```python
from langchain_openai import OpenAIEmbeddings
modelName = 'text-embedding-3-small'
embeddings = OpenAIEmbeddings(model = modelName)
embeddings
```

```python
sentence_embeddings = embeddings.embed_documents(sentences)
sentence_embeddings
```

```python
## Calculate the Similarity between all the pairs
for i in range(len(sentences)):
    for j in range(i+1,len(sentences)):
        similarity = cosine_similarity(sentence_embeddings[i],sentence_embeddings[j])
        print(f"{sentences[i]} vs {sentences[j]}")
        print(f" Similarity: {similarity:.3f}\n")
```

```python
### Example2 - Semantic Search - Retrive the simialr sentence
# Test Sematic Search
documents = [
    "Artificial intelligence enables machines to perform tasks that normally require human intelligence.",
    "Machine learning allows computers to learn patterns from data without being explicitly programmed.",
    "Deep learning uses neural networks with multiple layers to solve complex problems.",
    "LangChain is a framework for building applications powered by large language models.",
    "LangChain can connect language models with external tools, databases, and APIs.",
    "Retrieval Augmented Generation helps language models generate answers using external knowledge.",
    "Vector databases store numerical representations of text called embeddings.",
    "Embeddings represent the semantic meaning of text as numerical vectors.",
    "A machine learning model learns relationships between input features and target variables.",
    "Supervised learning uses labeled data to train predictive models.",
    "Unsupervised learning discovers hidden patterns in unlabeled data.",
    "Reinforcement learning trains agents using rewards and penalties.",
    "Large language models can understand and generate human-like text.",
    "AI agents can use tools and reasoning to perform multi-step tasks.",
    "LangChain can be used to build RAG pipelines for question-answering systems."
]
query = "What is Langchain???"
```

```python
def semantic_search(query,docuemnts,embedding_model,top_k=3):
    """ Simple Sematic Search Implementation"""
    ## Embed Query and Documents

    query_embeddings = embedding_model.embed_query(query)
    doc_embeddings = embedding_model.embed_documents(docuemnts)


    ## Calculate the Similarity Score
    similarities = []
    for i,doc_emb in enumerate(doc_embeddings):
        similarity = cosine_similarity(query_embeddings,doc_emb)
        similarities.append({similarity,documents[i]})

    ## Sort By Similarity
    similarity.sort(reverse = True)
    return similarity[:top_k]
```

```python
results = semantic_search(query,documents,embeddings)
results
```

```python
print(f"\n Sematic Search Results")
for score,doc in results:
    print(f"Score: {score:.3f} | {doc}")
```
